## 1.3 Enrichment

```
stats=df.groupby('clinic_id').agg(n_appointments=('appointment_id','count'),attendance_rate=('attended','mean'),no_show_rate=('no_show','mean'),avg_distance_km=('distance_km','mean'),avg_lead_days=('lead_days','mean'),avg_age=('age','mean')).reset_index()
enriched=clinics.merge(stats,on='clinic_id',how='left'); enriched.to_csv('clinics_enriched.csv',index=False); enriched
```
